# 🧠 Single Agent Pipeline Project

## Problem Statement
Build a **Single-Agent Smart Assistant** that:
- Understands user queries
- Routes tasks based on intent
- Uses tools when required
- Returns structured JSON output

### The agent should handle:
- Math queries → Calculator Tool
- Keyword extraction → Keyword Tool
- General queries → Direct response

````python
# 🛠️ TOOL 1: Calculator

import re
def calculator(expression: str) -> str:
    try:
        expression = expression.replace(" ", "")
        if not re.fullmatch(r"[0-9+\-*/().%\s]+", expression):
            return "Error: Invalid characters in expression."

        result = eval(expression, {"__builtins__": None}, {})
        return str(result)
    except ZeroDivisionError:
        return "Error: Division by zero."

    except SyntaxError:
        return "Error: Invalid mathematical expression."

    except Exception:
        return "Error in calculation."
````

````python
# 🛠️ TOOL 2: Keyword Extractor
import re

def extract_keywords(text: str) -> list:
    try:
        stopwords = {
            "the", "is", "are", "was", "were", "this", "that",
            "have", "has", "had", "with", "from", "into", "about",
            "would", "could", "should", "there", "their", "which",
            "where", "when", "what", "your", "than", "then",
            "been", "being", "they", "them", "will", "shall",
            "and", "for", "you", "our", "not", "but", "can"
        }
        words = re.findall(r"\b[a-zA-Z]+\b", text.lower())
        keywords = []
        seen = set()
        for word in words:
            if len(word) > 4 and word not in stopwords and word not in seen:
                keywords.append(word)
                seen.add(word)
        return keywords[:5]
    except Exception:
        return []

````

````python
# 🛠️ TOOL 3 : Word Counter

def count_words(text: str) -> int:
    """Count the number of words in the given text."""
    try:
        return len(text.split())
    except Exception:
        return 0
````

## 🤖 Implement Agent Logic Below

👉 Use conditional routing:
- If query contains "calculate" → use calculator
- If query contains "keywords" → use keyword extractor
- Else → general response

````python
# 🤖 AGENT FUNCTION (TO IMPLEMENT)

import re
import logging

logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)


def agent(query: str):
    """
    Routes a user query to the appropriate tool based on intent.

    Returns:
        {
            "type": "calculation | keywords | word_count | general | error",
            "result": ...
        }
    """

    if not isinstance(query, str) or not query.strip():
        logger.warning("Received empty or invalid query.")
        return {
            "type": "error",
            "result": "Query must be a non-empty string."
        }

    query_lower = query.lower()
    logger.info(f"Received query: {query}")

    try:
        if "calculate" in query_lower:

            logger.info("Routing to Calculator Tool")

            expression = query_lower.replace("calculate", "").strip()
            expression = re.sub(r"[^0-9+\-*/().% ]", "", expression)

            if not expression:
                return {
                    "type": "error",
                    "result": "No valid mathematical expression found."
                }

            result = calculator(expression)

            return {
                "type": "calculation",
                "result": result
            }
        elif "keyword" in query_lower or "keywords" in query_lower:

            logger.info("Routing to Keyword Tool")

            keywords = extract_keywords(query)

            return {
                "type": "keywords",
                "result": keywords
            }

        elif "count" in query_lower and "word" in query_lower:

            logger.info("Routing to Word Count Tool")

            result = count_words(query)

            return {
                "type": "word_count",
                "result": result
            }

        else:

            logger.info("General Query")

            return {
                "type": "general",
                "result": f"You asked: '{query}'. No specific tool was required."
            }

    except Exception as e:

        logger.error(f"Agent failed: {e}")

        return {
            "type": "error",
            "result": f"Something went wrong: {str(e)}"
        }
print(agent("Calculate 25 + 10"))

print(agent("Extract keywords from Artificial Intelligence is transforming healthcare"))

print(agent("Count the words in this sentence"))

print(agent("Who is Virat Kohli?"))
````
